Copy List 

```python
mylist = ['one', 'two', 'three', 'four', 'five', 'six', 'seven', 'eight', 'nine']
mylist
```

```python
# Create a new reference "mylist1"
mylist1 = mylist
mylist1
```

```python
 # The address of both mylist & mylist1 will be different 
print(id(mylist1))
print(id(mylist))
```

```python
#Create a copy of the list 
mylist2 = mylist1.copy()
```

```python
mylist2
```

```python
id(mylist2)
```

```python
mylist[0]=1
```

```python
mylist
```

```python
# mylist1 will be also impacted as it is pointing to the same list 
mylist1
```

```python
# Copy of list won't be impacted due to changes made on the original list
mylist2
```
