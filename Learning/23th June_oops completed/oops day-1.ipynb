```python
class computer():        
    
    def __init__(self,cpu,ram): # cpu & ram as argument
        self.cpu = cpu #hear self is the object & we need to refer to the object
        self.ram = ram 
  
    def config(self):     
        print('config is:', self.cpu, self.ram)

com1 = computer('i5', 8)
com2 = computer('i7', 16)


com1.config() 
com2.config()
```

```python
class computer():        
    
    def __init__(self,cpu,ram,price,brand): # cpu & ram as argument
        self.cpu = cpu #hear self is the object & we need to refer to the object
        self.ram = ram 
        self.price = price
        self.brand = brand
  
    def config(self):     
        print('config is:', self.cpu, self.ram, self.price, self. brand)

com1 = computer('i5', 8, 50, 'hp')
com2 = computer('i7', 16, 30, 'lenovo')


com1.config() 
com2.config()
```

# constructor how it works 

```python
class computer():
    
    def __init__(self):
        self.name = 'akash'
        self.age = 37

c1 = computer()
c2 = computer()

print(c1.name,c1.age)
print(c2.name)
```

```python
class computer:
    def __init__(self):
        self.name = 'akash'
        self.age = 37

c1 = computer()
c2 = computer()

c2.name = 'mit'
c1.name = 'rakesh'

print(c1.name)
print(c2.name)
```

```python
class computer():
    def __init__(self):
        self.name = 'akash'
        self.age = 37

c1 = computer()
c2 = computer()

c1.name = 'nit'
c1.age = 20

print(c1.name, c1.age)
print(c1.age)
print(c2.name)

print(id(c1))
print(id(c2))
```
