```python
class student(): 
        
    school = 'nit'
    
    def __init__(self, m1, m2, m3) : 
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    
    def avg(self):   # ------> hear this average is INSTANCE METHOD 
        return (self.m1+self.m2+self.m3)/3

    @classmethod
    def info(cls): # uer need to define cls - for class variable 
        return cls.school 

    @staticmethod
    def bye():
        return ('static method')
        
# lets create  object hear as we created s1, s2 we created 2 object & also i want to pass value in this  
s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

print(s1.avg()) # instance method cuz it acess instance variabble, m1, m2, m3 
print(s1.info()) # class method cuz we pass argument cls, cls varible school 
print(s1.bye()) # static method cuz it wont access any variable 
```
