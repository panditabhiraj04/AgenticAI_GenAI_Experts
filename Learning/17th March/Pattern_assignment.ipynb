```python
# Right angle tringle pattern    range excluide end element
for i in range(1,6):
    print('* '*i)
```

```python
#Inverted right angle triangle 
for i in range(1,6):
    print('* '*(6-i))

# 2nd method 
for i in range(5,0,-1):
    print('* '*i)
```

```python
# pyramid pattern 
for i in range(1,10,2):
    print('* '*i)

```

```python
# inverted pyramid pattern 
for i in range(1,10,2):
    print('* '*(10-i))

#2nd method 
for i in range(10,1,-2):
    print('* '*(i-1))
```

```python
#Diamond Pattern
for i in range(1,10,2):
    print('* '*i)
    if i == 9 :
        for j in range(i,1,-2):
            print('* '*(j-2))
    
    
```

```python
# hallow squar pattern
for i in range(1,6):
    if i == 1 or i ==5:
        print('*'*5)
    if i == 2 or i==3:
        for j in range(i,1,-1):
            print('*   *')

print('')
# hallow squar pattern

for i in range(5):
    for j in range(5):
        if  i == 0 or i == 4 or j == 0 or j ==4:
            print('*', end ='')
        else:
            print(' ', end = '')

    print('')
    
```
